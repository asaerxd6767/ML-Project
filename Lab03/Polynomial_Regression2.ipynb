```python
import pandas as pd 
df = pd.read_csv("insurance.csv")
df.head()
```

### Data Preparation

```python
df.isna().sum()
```

```python
df['region'].unique()
```

```python
df['sex'] = df['sex'].replace({'female': 1, 'male': 2})
df['smoker'] = df['smoker'].replace({'yes': 1, 'no': 2})
df['region'] = df['region'].replace({'southwest': 1, 'southeast': 2, 'northwest': 3, 'northeast': 4})
```

### Defining Input Features and Target Variable

```python
X = df.drop(columns = ['charges'])
y = df['charges']
```

```python
X
```

```python
y
```

### Split for Training and Testing

```python
from sklearn.model_selection import train_test_split 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state = 1)
```

### Scaling the Data

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaler = scaler.fit_transform(X_train)
X_test_scaler = scaler.transform(X_test)
```

### Model Development and Evaluation

```python
from sklearn.linear_model import LinearRegression

from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=3)

poly_features_xtrain = poly.fit_transform(X_train_scaler)

poly_features_xtest = poly.transform(X_test_scaler)

lin = LinearRegression()

lin.fit(poly_features_xtrain ,  y_train)




```

```python
y_pred = lin.predict(poly_features_xtest)
```

```python
from sklearn.metrics import mean_absolute_error  #error of test
mean_absolute_error(y_test, y_pred)
```

```python
y_pred_train = lin.predict(poly_features_xtrain)
```

```python
mean_absolute_error(y_train, y_pred_train) #error of rain
```

```python
from sklearn.metrics import r2_score

r2_test = r2_score(y_test, y_pred)
r2_train = r2_score(y_train, y_pred_train)

print("R2 Test Score:", r2_test)
print("R2 Train Score:", r2_train)
```
